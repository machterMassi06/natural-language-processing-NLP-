# NER (NAME ENTITY RECOGNITION)

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
nlp.pipe_names
```

Let's discover NER with a simple example

```python
doc = nlp("Tesla Inc is going to acquire twitter for $45 billion")

for ent in doc.ents:
  print(ent.text, "|", ent.label_, "|", spacy.explain(ent.label_))
```

We can also change the display to make very visible our entities

```python

from spacy import displacy

displacy.render(doc, style="ent")
```

# Let's practise with some exercises

## Exercise 1: Extract Geographical Names

**Task:** Extract all geographical names (cities, countries, states) from the given text. The NER label for this is `GPE` (Geopolitical Entity).

```python
text_geo_en = "Paris, the capital of France, is located in Europe. New York City in the United States is a global hub, while Sydney is a major city in Australia."
doc_geo_en = nlp(text_geo_en)

# Your code here to extract geographical entities (GPE)

geo_entities = [
    ent.text
    for ent in doc_geo_en.ents
    if ent.label_ == "GPE"
]
geo_entities
```

## Exercise 2: Extract Cricketers' Birth Dates

**Task:** Extract all birth dates mentioned for cricketers in the given text. The NER label for dates is `DATE` and for persons is `PERSON`.

```python
text_en = "Jean was born on November 5, 1988. Bob birthday is April 24, 1973. Jack, a famous teacher, was born on May 2, 1969."
doc_en = nlp(text_en)

# Your code here to extract dates and player names (PERSON)
# You might need to combine information about PERSON and DATE.

dates = [
    ent.text
    for ent in doc_en.ents
    if ent.label_ == "DATE"
]

persons = [
    ent.text
    for ent in doc_en.ents
    if ent.label_ == "PERSON"
]

print(dates)
print(persons)
person_date = zip(persons, dates)
print()
for person, date in person_date:
  print(f"{person} was born on {date}")

```

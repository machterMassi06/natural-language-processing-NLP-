```python
import spacy
```

# 1. Create a blank English pipeline
the blank("en") pipeline contains the English tokenizer, but it does not contain a statistical model.

```python
nlp = spacy.blank("en")
type(nlp)
```

#2. Basic tokenization
Let's start with a simple sentence: spaCy automatically separates words and punctuation into different tokens.

```python
text = "Hello, my name is John."
doc = nlp(text)
for token in doc: print(token.text)
```

```python
# We can easily convert the tokens into a Python list
token_list = [token.text for token in doc]
print(token_list)
```

# 3. Token attributes
Each token contains useful information.

```python
text = "I have 2 apples."
doc = nlp(text)
for token in doc: print(f"token : {token.text}  --  token_alpha ? : {token.is_alpha}, token_digits ?: {token.is_digit}")
```

We can inspect several properties at the same time:

```python
for token in doc:
  print( token.text, " | alpha:", token.is_alpha, " | digit:", token.is_digit, " | punctuation:", token.is_punct, " | space:", token.is_space )
```

```python
text2 = "let's go to N.Y.!"
doc2 = nlp(text2)
for token in doc2:
  print(token)
```

# 5. Application : Text Processing Task
So far, we have worked with short sentences. Now let's work with a more realistic piece of text.

The goal is to take a raw text and progressively transform it into a clean list of words.

```python
text = """ Artificial intelligence is changing the way we work, learn, and communicate. In recent years, tools such as ChatGPT have become increasingly popular among students, researchers, and professionals. However, using AI effectively is not just about asking questions; it is also about understanding how these systems process information. For example, when we give a computer a sentence, the first step in many NLP applications is tokenization. The text must be divided into smaller units, called tokens. These tokens can then be cleaned, analyzed, and transformed before being used by a machine learning model. But what happens when the text contains punctuation, numbers, or special characters? Consider this example: "NLP is amazing!" A computer does not necessarily need the comma, the quotation marks, or the exclamation mark when we want to analyze the words themselves. """
```

```python
doc = nlp(text)
for token in doc : print(token)
```

```python
print("#tokens: ", len(doc))
```

```python
from catalogue import check_exists
#Create a list containing all the tokens
tokens = [token.text for token in doc]

# cleaned tokens (lowercases,remove puncts, and keep only alphas)
cleaned_tokens = [
    token.text.lower()
    for token in doc
    if not token.is_punct and token.is_alpha
]

# How many words are there
print("Original tokens:", len(tokens))
print("Clean words:", len(cleaned_tokens))

# find unique words
print("unique words: ",len(set(cleaned_tokens)))
# display some of them
print("Some of unique words: ",list(set(cleaned_tokens))[:20])

# words frequency - We can count how often each word appears

from collections import Counter

w_frequency = Counter(cleaned_tokens)
print("word frequency/most_common 10: ", w_frequency.most_common(10))
```

Conclusion : This is a simple example of a text preprocessing pipeline, which is often used as an early step (preprocessing text) before further NLP tasks such as text classification, sentiment analysis, information retrieval, or machine learning.

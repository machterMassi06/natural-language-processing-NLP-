## Spacy NLP Pipeline Basics

This notebook demonstrates fundamental concepts of using the spaCy library for Natural Language Processing (NLP).

### 1. Importing spaCy

First, we import the `spacy` library, which is essential for all NLP operations within this framework.

```python
import spacy
```

### 2. Blank Pipeline

A blank spaCy pipeline (`spacy.blank("en")`) provides only a tokenizer. It can split text into tokens but doesn't perform any further linguistic analysis like part-of-speech tagging or named entity recognition, as it has no loaded components.

```python
nlp = spacy.blank("en") # blank pipeline
nlp.pipe_names # empty pipeline
```

## 3. Sample Text

We define a sample text string that will be processed using different spaCy pipelines. This text will be the input for our NLP tasks.

```python
text="Hello from france. I'am learning nlp pipeline with spacy!."
```

### 4. Tokenization with Blank Pipeline

Using the blank pipeline, we process the text to create a `Doc` object. We then iterate through the tokens in this `Doc` object and print each token. This demonstrates basic tokenization.

```python
doc = nlp(text)
for token in doc: print(token.text)
```

### 5. Loading a Pre-trained Model

To perform more advanced NLP tasks, we load a pre-trained English model (`en_core_web_sm`). This model comes with various pipeline components (e.g., tagger, parser, ner) that enable richer linguistic analysis beyond simple tokenization. The `nlp.pipe_names` attribute shows all components loaded in the pipeline.

```python
nlp = spacy.load("en_core_web_sm")
nlp.pipe_names # all components
```

### 6. Advanced Token Analysis

With the loaded pre-trained model, we can now access various attributes of each token, such as its part-of-speech tag (`token.pos_`) and its lemma (`token.lemma_`). The lemma is the base form of a word, and `pos_` provides the grammatical category of the word.

```python
doc = nlp(text)
for token in doc: print(token.text, " | ", token.pos_, " | ", token.lemma_)
```

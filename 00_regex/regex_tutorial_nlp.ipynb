# NLP Tutorial: Regular Expressions

## 1. Import Libraries

```python
import re
```

## 2. Customer Chat Data: Order Number Related

```python
chat1='Costumer: Hello, I am having an issue with my order # 412889912'
chat2='Costumer: I have a problem with my order number 412889912'
chat3='Costumer: My order 412889912 is having an issue, I was charged 300$ when online it says 280$'
```

## 3. Define Regex for Order Numbers

```python
#Regex to Retrieve order number
pattern = r'order[^\d]*(\d+)'
```

## 4. Extracting Order Numbers from Chat

```python
print(re.findall(pattern,chat1))
print(re.findall(pattern,chat2))
print(re.findall(pattern,chat3))
```

## 5. Customer Chat Data: Contact Information Related

```python
chat1 = 'Costumer: you ask lot of questions 😠  9876543210, new_email_1@example.com'
chat2 = 'Costumer: here it is: (987)-654-3210, new_email_2@example.com'
chat3 = 'Costumer: yes, phone: 9876543210 email: new_email_3@example.com'
```

```python
def get_match(pattern,text):
  """
  Extracts the first matching group for a given regex pattern from a text.
  """
  matches = re.findall(pattern,text)
  if matches:
    return matches[0]
```

```python
get_match(r'[a-zA-Z0-9_]*@[a-z]*.[a-zA-Z0-9]*',chat1)
```

```python
get_match(r'[a-zA-Z0-9_]*@[a-z]*.[a-zA-Z0-9]*',chat2)
```

```python
get_match(r'[a-zA-Z0-9_]*@[a-z]*.[a-zA-Z0-9]*',chat3)
```

### Phone number

```python
get_match(r'\d{10}|\(\d{3}\)-\d{3}-\d{4}',chat1)
```

```python
get_match(r'\d{10}|\(\d{3}\)-\d{3}-\d{4}',chat2)
```

```python
get_match(r'\d{10}|\(\d{3}\)-\d{3}-\d{4}',chat3)
```

## 6. Information Extraction

```python
text='''
Born	Elon Reeve Musk
June 28, 1971 (age 50)
Pretoria, Transvaal, South Africa
Citizenship
South Africa (1971–present)
Canada (1971–present)
United States (2002–present)
Education	University of Pennsylvania (BS, BA)
Title
Founder, CEO and Chief Engineer of SpaceX
CEO and product architect of Tesla, Inc.
Founder of The Boring Company and X.com (now part of PayPal)
Co-founder of Neuralink, OpenAI, and Zip2
Spouse(s)
Justine Wilson
​
​(m. 2000; div. 2008)​
Talulah Riley
​
​(m. 2010; div. 2012)​
​
​(m. 2013; div. 2016)
'''
```

```python
get_match(r'age (\d+)',text)
```

```python
get_match(r'Born(.*)\n', text).strip()
```

```python
get_match(r'Born.*\n(.*)\(age', text).strip()
```

```python
def extract_personal_information(text):
    age = get_match(r'age (\d+)', text)
    full_name = get_match(r'Born(.*)\n', text)
    birth_date = get_match(r'Born.*\n(.*)\(age', text)
    birth_place = get_match(r'\(age.*\n(.*)', text)

    return {
        'age': int(age),
        'name': full_name.strip(),
        'birth_date': birth_date.strip(),
        'birth_place': birth_place.strip()
    }
```

```python
extract_personal_information(text)
```

# Stop Words in NLP

## What are stop words?

Stop words are very common words in a language that usually do not carry much meaning by themselves, such as "the", "is", "and", "of", or "in". In many NLP tasks, these words are removed because they appear very frequently and do not help much to distinguish between documents or sentences.

## Why are stop words important in NLP?

They are often filtered out during preprocessing to reduce noise and make text data easier to analyze. By removing frequent but low-information words, models can focus on more meaningful terms and work more efficiently. This is especially useful in text classification, search engines, topic modeling, and information retrieval.

## Advantages of removing stop words

- Reduces the size of the vocabulary.
- Speeds up training and processing.
- Lowers noise in the data.
- Improves the performance of some basic NLP models.

## Disadvantages of removing stop words

- Important words such as "not", "no", or "never" can be lost.
- This can harm sentiment analysis and phrase understanding.
- Some tasks, such as chatbots or question answering, need these words to keep the correct meaning.

In short, stop words are useful in many cases, but they should not always be removed blindly because context and task type matter a lot in NLP.

```python
import spacy
from spacy.lang.en.stop_words import STOP_WORDS
```

```python
print(f"Stops words defined in spaCy: ", STOP_WORDS)
```

```python
len(STOP_WORDS)
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("We just opened our wings, the flying part is coming soon")

for token in doc :
  if token.is_stop: print(f"Stop word found: {token.text}")
```

```python
def preprocess_text(text):
  """
  Return the preprocessed text (String without stop words)
  """
  return " ".join([token.text for token in nlp(text) if not token.is_stop])
```

```python
preprocess_text("We just opened our wings, the flying part is coming soon")
```

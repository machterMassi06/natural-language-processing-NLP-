# Stemming in NLTK

### Stemming
Stemming is a technique used in natural language processing (NLP) to reduce inflected (or sometimes derived) words to their word stem, base or root form—generally a written word form. The stem it may not be a word at all

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
```

```python
words = ["eating", "eats", "eat", "ate", "adjustable", "rafting", "ability", "meeting", "better"]
```

```python
for word in words:
  print(word, ":", stemmer.stem(word))
```

# Lemmatization in Spacy

### Lemmatization
Lemmatization, in contrast to stemming, is an algorithmic process of determining the lemma of a word based on its meaning. It involves reducing inflected words to their dictionary form (lemma). For example, 'better' is lemmatized to 'well'.

```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm") # smale prentrined "en" model
```

```python
doc = nlp("eating eats eat ate adjustable rafting ability meeting better")
for token in doc:
    print(token, " | ", token.lemma_)
```

## Spacy : Customizing lemmatizer

```python
doc = nlp("Bro, you wanna go? Brah, don't say no! I am exhausted")
```

```python
for token in doc:
    print(token.text, "|", token.lemma_)
```

Bro, brah ?? all mean brother. so let's costomise one rule

```python
nlp.pipe_names
```

```python
ar = nlp.get_pipe('attribute_ruler')

ar.add([[{"TEXT":"Bro"}],[{"TEXT":"Brah"}]],{"LEMMA":"Brother"})
doc = nlp("Bro, you wanna go? Brah, don't say no! I am exhausted")
for token in doc :
  print(token.text , ":", token.lemma_)
```

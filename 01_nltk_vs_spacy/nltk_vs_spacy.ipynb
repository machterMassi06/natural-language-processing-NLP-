## NLTK vs. spaCy

NLTK (Natural Language Toolkit) is a powerful library for natural language processing, offering a wide range of algorithms for tasks like tokenization, stemming, tagging, parsing, and classification. It's often used for research and educational purposes due to its comprehensive nature.

spaCy is a more modern and efficient NLP library designed for production use. It focuses on providing fast and accurate pre-trained models for various NLP tasks, including named entity recognition, dependency parsing, and text classification. We will quickly explore their differences in this notebook.

```python
import spacy
```

```python
# load specifid package
# load english core
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp("Apple is looking at buying U.K. startup for $1 billion. Apple has a lot of money.")
```

```python
for sentence in doc.sents:
  print(sentence)
```

```python
for sentence in doc.sents:
  for w in sentence:
    print(w)
```

```python
import nltk
from nltk.tokenize import sent_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
texte = "Apple is looking at buying U.K. startup for $1 billion. Apple has a lot of money."

sentences = sent_tokenize(texte)

print(sentences)
```

```python
from nltk.tokenize import word_tokenize

words = word_tokenize(texte)

print(words)
```
